```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

loader = PyPDFLoader("RAG/notebook/sample_pdfs/sample.pdf")
docs = loader.load()
print(docs)
```

```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_community.document_loaders import DirectoryLoader
loader = DirectoryLoader(
    "RAG/notebook/sample_pdfs",
     glob="**/*.pdf", #loads all the pdf files in this and all other locations specified aswell
     loader_cls=PyPDFLoader) #selects which loader to use

docs = loader.load()
print(docs)
```
