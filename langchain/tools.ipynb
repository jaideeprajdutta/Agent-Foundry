```python
import os
from langchain.chat_models import init_chat_model

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash-lite")
response = model.invoke("What are movies about AI?")
print(response.content)
```

```python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
    """Get the weather for a given location."""
    return f"The weather in {location} is sunny."


model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What is weather in hyderabad")
print(response.tool_calls)
```

###Tool Execution using loops

```python
messages = [{"role": "user", "content": "What is the weather in hyderabad?"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)
print(messages)


#Execute tools and collect results

for tool_call in ai_msg.tool_calls:
    
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)

final_response = model_with_tools.invoke(messages)
print(final_response.content)
```

```python
messages = [{"role": "user", "content": "What is the weather in hyderabad mumbai and delhi?"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)
print(messages)


#Execute tools and collect results

for tool_call in ai_msg.tool_calls:
    
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)

final_response = model_with_tools.invoke(messages)
print(final_response.content)
```

```python
from langchain_core.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get the weather for a given city."""
    return f"The weather in {city} is sunny."

tool_calls = [
    {"city": "hyderabad"},
    {"city": "mumbai"},
    {"city": "delhi"}
]

messages = []

for i in tool_calls:
    tool_result = get_weather.invoke(i)
    messages.append(tool_result)

print(messages)
```
