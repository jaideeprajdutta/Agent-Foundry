```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")

model = init_chat_model("google_genai:gemini-2.5-flash")
response = model.invoke("What are movies about AI?")
print(response.content)
```

```python
for i in model.stream("What are movies about AI?"):
    print(i.content, end="")
```

```python
responses = model.batch(["What are movies about AI?", "What are books about AI?" , "Tell me about best authors and directors"])

for response in responses:
    print(response.content)
```
