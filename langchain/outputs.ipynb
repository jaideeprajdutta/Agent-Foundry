```python
import os
from langchain.chat_models import init_chat_model
os.environ["OPENROUTER_API_KEY"]=os.getenv("OPENROUTER_API_KEY")
model=init_chat_model("openrouter:qwen/qwen3-32b")

response = model.invoke("hello, how are you")
print(response)
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title:str = Field(description="The title of the movie")
    director:str = Field(description="The director of the movie")
    year:int = Field(description="The year the movie was released")
    rating:float = Field(description="The rating of the movie")
```

```python
model_with_structured = model.with_structured_output(Movie)
model_with_structured.invoke("give me details about the movie Kill Bill")
```

```python
response = model_with_structured.invoke("tell me about movie pulp fiction")
print(response)
```

Nested Structure

```python
from pydantic import BaseModel, Field

class Director(BaseModel):
    name: str = Field(description="The name of the director")
    age: int = Field(description="The age of the director")

class MovieDetails(BaseModel):
    title: str = Field(description="The title of the movie")
    director: Director = Field(description="The director of the movie")
    year: int = Field(description="The year the movie was released")
    rating: float = Field(description="The rating of the movie")

model_with_structure = model.with_structured_output(MovieDetails)

response = model_with_structure.invoke("Tell me about Dark Knight")
print(response)
```

```python
from pydantic import BaseModel, Field
from langchain.agents import create_agent

class ContactInfo(BaseModel):
    """Contact infor for a person"""
    name: str = Field(description="The name of the person")
    email: str = Field(description="The email of the person")
    phone: str = Field(description="The phone number of the person")

agent = create_agent(
    model = "openrouter:qwen/qwen3-32b",
    response_format = ContactInfo
)

response = agent.invoke({"messages": [{"role": "user", "content": "give info or extract from : jaideep , jaideep@gmail.com , 1234567890"}]})
print(response)
```
